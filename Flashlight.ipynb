# Flashlight evidence walkthrough

One case, general wording first. Set the external paths below using the README.

```python
import json, sys
sys.dont_write_bytecode = True
from walkthrough import run
source = 'path/to/Flashlight Example.sysml'
pilot = 'path/to/pilot/sysml'
scratch = 'path/to/external/scratch'
encoder = None
report = json.loads(run(source, pilot, scratch, encoder).read_text(encoding='utf-8'))
```

```python
for row in report['conditions']:
    print(row['question'], row['method'], row['evidence_tokens'], row['pass'])
    for fact, audit in row['facts'].items():
        print(fact, audit['status'], audit['candidate_ranks'])
```

```python
row = report['conditions'][0]  # Choose 0–5
packed = {item['id'] for item in row['packed']}
for rank, record in enumerate(row['ranked'], 1):
    print(rank, 'packed' if record['id'] in packed else 'omitted', record['text'], sep='\n')
```
